```python
# Example list of locations (latitude, longitude, name)
locations = [
 #   (53.34803180330561, -6.275358804387943, "Dublin"),     # Dublin
  #  (51.88940626480707, -8.490771460163085, "Cork"),       # Cork
   # (53.28567307231417, -9.028553794056856, "Galway"),     # Galway

(53.4, -6.2, "Dublin"),
(51.8, 	-8.3, "Cork"),
(53.4, 	-9.0, "Galway"),
(52.6,	-8.6, "Limerick"),
(52.3,	-7.1, "Waterford"),
(54.3, 	-8.5, "Sligo"),
(52.5,	-8.2, "Tipperary"),

]
```

```python
import folium
import base64
from IPython.display import display, IFrame

def display_esri_map_inline(locations):
    if len(locations) > 0:
        initial_location = locations[0][:2]

        # Use ESRI World Street Map tiles (free, public, and no API key needed)
        esri_tiles_url = 'https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}'
        esri_attribution = 'Tiles &copy; Esri &mdash; Source: Esri, DeLorme, NAVTEQ, USGS, Intermap, iPC, and the GIS User Community'

        m = folium.Map(
            location=initial_location,
            zoom_start=6,
            tiles=esri_tiles_url,
            attr=esri_attribution
        )

        for lat, lon, name in locations:
            if name == "Dublin":
                folium.Marker(location=[lat, lon], popup=name,
                              icon=folium.Icon(color='red')).add_to(m)
            else:
                folium.Marker(location=[lat, lon], popup=name).add_to(m)

        # Convert to string and embed inside a secure Base64 IFrame to display inside Colab securely
        map_html = m._repr_html_()
        b64_html = base64.b64encode(map_html.encode('utf-8')).decode('utf-8')
        data_uri = f"data:text/html;charset=utf-8;base64,{b64_html}"

        print("Rendering interactive ESRI map below...")
        display(IFrame(src=data_uri, width='100%', height=500))
    else:
        print("No locations provided.")

# Display the map
display_esri_map_inline(locations)
```
